```python
def addition( a, b):
    return a + b
def subtraction( a, b):
    return a - b
def multiplication( a,  b):
    return a * b
def division( a,  b):
    if b == 0:
        return "Error: Division by zero is not allowed."
    return a / b
```

```python
print("Welcome to the calculator!")
start = ""
while start != "exit":
    s = input("Enter the operation (addition, subtraction, multiplication, division):")
    num1 = float(input("Enter first number:"))
    num2 = float(input("Enter second number:"))
    if s == "addition":
        print(addition(num1, num2))
    elif s == "subtraction":
        print(subtraction(num1, num2))
    elif s == "multiplication":
        print(multiplication(num1, num2))
    elif s == "division":
        print(division(num1, num2))
    else:
        print("Invalid operation. Please try again.")
    end = input("Type 'e' to quit or 'c' to continue: ")
    if end == 'e':
        start = "exit"
    elif end == 'c':
        continue
    else:        
        print("Invalid input. Continuing with the calculator.")
```
