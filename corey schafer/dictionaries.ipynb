```python
# creating dictionary
my_dict = {"name":"alice","age":30,"city":"new york"}
```

```python
print(my_dict)
```

```python
print(my_dict['age'])
```

```python
result = [my_dict.get(k) for k in ['name', 'age']]
print(result)
```

```python
my_dict.get(list(my_dict.keys())[0])
```

```python
print(my_dict.keys())
print(my_dict.values())
```

```python
my_dict.items()
```

```python
my_dict['country'] = 'usa'
```

```python
# how to add a new key value pair to a dictionary
#to add a new key value pair to a dictionary, you can simply assign a value to a new key using the following syntax:
my_dict['new_key'] = 'new_value'
```

```python
#how to update a value in a dictionary
# to update a value in a dictionary, you can assign a new value to an existing key using the following syntax:
my_dict['name'] = 'bob'
```

```python
# how to delete a key value pair from a dictionary
# to delete a key value pair from a dictionary, you can use the del statement followed by the key you want to delete. For example:
del my_dict['age']
```

```python
# dictionary comprehension
squared_dict = {x: x**2 for x in range(6)}
```

```python
print(squared_dict)
```

```python
# nested dictionaries
nested_dict = {
    'dict1': {'name': 'alice','age': 30},
    'dict2': {'name': 'bob','age': 25},
    'dict3': {'name': 'charlie','age': 35}
}
```

```python
print(nested_dict)
```
