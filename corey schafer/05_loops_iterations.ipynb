```python
nums = [1,2,3,4,5,6,7]
```

```python
for i in nums:
    print(i**3)
```

```python
for num in nums:
    if num == 3:
        print('Found!')
        break
    print(num)
```

```python
for num in nums:
    if num == 3:
        print('Found!')
        continue
    print(num)
```

```python
for num in nums:
    for letter in 'abc':
        print(num, letter)
```
