```python
import random
```

```python
print("welcome to number guessing game.")
print("Enter your range for the number to be guessed.")
r1=int(input("Enter your starting number(smaller): "))
r2=int(input("Enter your ending number(larger): "))
if r1>r2:
    print("Invalid range. Starting number should be smaller than ending number.")
    exit()
    
guesses=int(input(f"Enter the number of attempts you want to guess the number between {r1} and {r2}: "))
print(f"You have {guesses} attempts to guess the number between {r1} and {r2}. Good luck!")
rand_num = random.randint(r1,r2)
print("The number has been generated. Start guessing!")
for i in range(guesses):
    num = int(input("Enter your guess: "))
    if num < r1 or num > r2:
        print(f"Invalid guess. Please enter a number between {r1} and {r2}.")
        break
    if num == rand_num:
        print(f"Congratulations! You guessed the number correctly {rand_num}.")
        break
    elif num < rand_num:
        print(f"{num} Too low! Try again.")
    else:
        print(f"{num} Too high! Try again.")
else:  
    print(f"Sorry, you've used all your attempts. The correct number was {rand_num}.")
```
