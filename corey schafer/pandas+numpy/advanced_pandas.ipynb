```python
df = pd.DataFrame({
    "Order_ID":[101,102,103,104,105,106,107,108,109,110],
    "Customer":["Amit","Priya","Rahul","Amit","Sneha",
                "Rahul","Neha","Priya","Amit","Sneha"],
    "City":["Mumbai","Pune","Delhi","Mumbai","Pune",
            "Delhi","Nagpur","Pune","Mumbai","Delhi"],
    "Category":["Electronics","Clothing","Electronics",
                "Furniture","Clothing","Electronics",
                "Furniture","Clothing","Electronics","Furniture"],
    "Amount":[5000,2000,np.nan,7000,2500,6000,np.nan,3000,8000,4500],
    "Quantity":[2,1,3,np.nan,2,
                2,1,np.nan,4,2]
})
```

```python
df["Amount"]=df["Amount"].fillna(0)
df
```

```python
mean_amount = df["Quantity"].mean()
df["Quantity"] = df["Quantity"].fillna(mean_amount)
df
```

```python
df["Amount"] = df["Amount"].ffill() # forward fill filles last value
df["Amount"] = df["Amount"].bfill() # backward fill filles last value
```

```python
df["Amount"] = df["Amount"].fillna(df["Amount"].median())

df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
```

```python
import pandas as pd

df = pd.DataFrame({
    "order_id":[101,102,103,104,105,106],
    "customer":["amit","priya","rahul","amit","sneha","vikas"],
    "amount":[5000,2000,3000,7000,2500,6000],
    "order_date":["2023-11-01","2023-06-21","2023-12-03",
                "2023-02-04","2023-01-05","2023-03-06"]
})

print(df)
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
print(df["order_date"])
```

```python
df["year"] = df["order_date"].dt.year
print(df[["order_date","year"]])
```

```python
df["month"] = df["order_date"].dt.month 
df["day"] = df["order_date"].dt.day
print(df[["order_date","month","day"]])
```

```python
df["month_name"] = df["order_date"].dt.month_name()
print(df[["order_date","month_name"]])
```

```python
df["quarter"] = df["order_date"].dt.quarter
print(df[["order_date","quarter"]])
```

```python
df["week"] = df["order_date"].dt.isocalendar().week
print(df[["order_date","week"]])
```

```python
df[df["order_date"] > "2023-02-01"]
```

```python
df["delivery_date"] = (
    df["order_date"] + pd.Timedelta(days=7)
)

print(df[["order_date","delivery_date"]])
```

```python
df["days_to_delivery"] = (df["delivery_date"] - df["order_date"])
print(df[["order_date","delivery_date","days_to_delivery"]])
```

```python
today = pd.Timestamp.today()
df["order_age"] = today - df["order_date"]
print(today)
print(df[["order_date","order_age"]])
```

```python
orders = pd.DataFrame({
    "Order_ID":[1,2,3,4,5,6],
    "Customer":["Amit","Priya","Rahul",
                "Amit","Sneha","Rahul"],
    "Category":["Mobile","Laptop","Mobile",
                "Laptop","Mobile","Laptop"],
    "Revenue":[5000,7000,3000,
               8000,4000,6000],
    "Order_Date":[
        "2025-01-10",
        "2025-01-20",
        "2025-02-05",
        "2025-02-15",
        "2025-03-10",
        "2025-03-22"
    ]
})

print(orders)
```
