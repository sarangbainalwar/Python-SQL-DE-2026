```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    "Order_ID":[101,102,103,104,105,106,107,108,109,110],
    "Customer":["Amit","Priya","Rahul","Amit","Sneha",
                "Rahul","Neha","Priya","Amit","Sneha"],
    "City":["Mumbai","Pune","Delhi","Mumbai","Pune",
            "Delhi","Nagpur","Pune","Mumbai","Delhi"],
    "Category":["Electronics","Clothing","Electronics",
                "Furniture","Clothing","Electronics",
                "Furniture","Clothing","Electronics","Furniture"],
    "Amount":[5000,2000,np.nan,7000,2500,6000,np.nan,3000,8000,4500],
    "Quantity":[2,1,3,np.nan,2,
                2,1,np.nan,4,2]
})
```

```python
df
```

```python
df.isnull()
```

```python
df["Amount"].isnull()
```

```python
df.isnull().sum()
```

```python
df.dropna()
```

```python
df.dropna(subset=["Amount"])
```

```python
df["Amount"]=df["Amount"].fillna(0)
df
```

```python
mean_amount = df["Quantity"].mean()
df["Quantity"] = df["Quantity"].fillna(mean_amount)
df
```

```python
df["Amount"] = df["Amount"].ffill() # forward fill filles last value
df["Amount"] = df["Amount"].bfill() # backward fill filles last value
```

```python
df["Amount"] = df["Amount"].fillna(df["Amount"].median())

df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
```

```python
df.isnull().sum()
```

```python
import pandas as pd

customers = pd.DataFrame({
    "customer_id":[1,2,3,4,5],
    "name":["Amit","Priya","Rahul","Sneha","Neha"],
    "city":["Mumbai","Pune","Delhi","Pune","Nagpur"]
})

customers
```

```python
orders = pd.DataFrame({
    "order_id":[101,102,103,104,105],
    "customer_id":[1,2,2,3,1],
    "Amount":[5000,2000,3000,4000,6000]
})

orders
```

```python
pd.merge(
    customers,
    orders,
    on="customer_id",
    how="inner"
)
```

```python
pd.merge(
    customers,
    orders,
    on="customer_id",
    how = "left"
)
```

```python
pd.merge(
    customers,
    orders,
    on="customer_id",
    how="right"
)
```

```python
pd.merge(
    customers,
    orders,
    on="customer_id",
    how="outer"
)
```

```python
customers=pd.DataFrame({
    "cust_id":[1,2,3]
})

orders=pd.DataFrame({
    "customer_id":[1,2,3]
})

pd.merge(
    customers,
    orders,
    left_on="cust_id",
    right_on="customer_id"
)
```

```python
df1 = pd.DataFrame({
    "name":["Amit","Priya"]
},index=[1,2])

df2 = pd.DataFrame(
    {"Salary":[50000,60000]},index=[1,2]
)

pd.merge(df1,df2,left_index=True,right_index=True)
```
