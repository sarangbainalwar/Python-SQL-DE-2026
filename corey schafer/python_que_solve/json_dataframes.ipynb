```python
# note Nested JSON — flatten address Medium Fetch /users. Each user has a nested address object. Extract city and zipcode from the nested dict and add them as flat columns in the DataFrame. Print the final DataFrame with name, city, zipcode.


import requests
import pandas as pd

url="https://jsonplaceholder.typicode.com/users"
res = requests.get(url)
```

```python
res.raise_for_status()
res=res.json()
df=pd.DataFrame(res)
```

```python
df.columns
```

```python
print(df.loc[0,"address"])
```

```python
df
```

```python
df["city"]=df["address"].apply(lambda x: x["city"])
```

```python
df["city"]
```

```python
df["zipcode"] = df["address"].apply(lambda x: x["zipcode"])
```

```python
df["zipcode"]
```

```python
print(df[["name","city","zipcode"]])
```

```python
# Deeply nested — geo coordinates Medium
# From the same /users endpoint, each address has a nested geo dict with lat and lng. Extract both into flat columns. Use pd.json_normalize() — not manual loops.

import requests
import pandas as pd

# Fetch data
url = "https://jsonplaceholder.typicode.com/users"
```

```python
users = requests.get(url).json()
```

```python
df=pd.json_normalize(users)
```

```python
df
```

```python
result = df [["name","address.geo.lat","address.geo.lng"]]

result = result.rename(columns={
    "address.geo.lat":"lat",
    "address.geo.lng":"lng"
})

print(result)
```

```python
# Join across two endpoints Medium Fetch /posts and /users separately. Merge them on userId / id so each post row also has the author's name and email. Print the first 5 rows showing title, name, email.

import requests
import pandas

posts = requests.get("https://jsonplaceholder.typicode.com/posts").json()
users = requests.get("https://jsonplaceholder.typicode.com/users").json()
```

```python
posts_df = pd.DataFrame(posts)
users_df = pd.DataFrame(users)
```

```python
merged_df = pd.merge(
    posts_df,
    users_df,
    left_on="userId",
    right_on="id",
    how="inner"
)
```

```python
print(merged_df[["title","name","email"]].head())
```
