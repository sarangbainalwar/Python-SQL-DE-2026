```python
# Dictionary comprehension — column mapping Medium
# The source schema uses snake_case. The target expects camelCase. Use a dict comprehension to transform the mapping and apply it to rename the columns list.

column_map = {
    "customer_id": "customerId",
    "order_date": "orderDate",
    "total_amount": "totalAmount",
    "is_refunded": "isRefunded"
}
columns = ["customer_id", "order_date", "product_name", "total_amount"]

# NOTE dict comprehension
mapping = {col: column_map[col] for col in columns if col in column_map}

# NOTE : Rename columns 
renamed_columns = [mapping.get(col,col) for col in columns]

print(mapping)
print(renamed_columns)
```

```python
# Nested dict — extract pipeline stats Medium
# Flatten the nested pipeline dict into a single-level dict with keys like sales_row_count, sales_status, inventory_row_count, etc.

pipelines = {
    "sales": {"row_count": 95000, "status": "success", "duration_sec": 120},
    "inventory": {"row_count": 12000, "status": "failed", "duration_sec": 45},
    "finance": {"row_count": 5400, "status": "success", "duration_sec": 200}
}

flat_dict = {}

for pipeline,stats in pipelines.items():
    print(pipeline,stats)
    for key,value in stats.items():
        print(key,value)
        flat_dict[f"{pipeline}_{key}"]=value 

print(flat_dict)
```
