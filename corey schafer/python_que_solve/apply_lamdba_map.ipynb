Classify salary bands Easy
Use .apply() with a lambda to add a salary_band column: "Low" if <60k, "Mid" if 60k–90k, "High" if >90k.
import pandas as pd
df = pd.DataFrame({
    "emp_id":  ["E01","E02","E03","E04","E05"],
    "name":    ["Arjun","Meera","Riya","Kiran","Sam"],
    "salary":  [85000, 58000, 65000, 92000, 71000]
})

```python
import pandas as pd

df=pd.DataFrame({
    "emp_id": ["E01", "E02", "E03", "E04", "E05"],
    "name": ["Arjun", "Meera", "Riya", "Kiran", "Sam"],
    "salary": [85000, 58000, 65000, 92000, 71000]
})

df["salary_band"] = df["salary"].apply(
    lambda x: "low" if x<60000 else ("mid" if x<=90000 else "high")
)

print(df)
```

Mask PII with map Easy
Use .map() on the email Series to mask the username — replace everything before @ with ***. Output: ***@accenture.com.
df = pd.DataFrame({
    "emp_id": ["E01","E02","E03"],
    "email":  ["arjun.k@accenture.com",
               "meera.r@gmail.com",
               "riya.s@accenture.com"]
})

```python
df=pd.DataFrame({
    "emp_id": ["E01", "E02", "E03"],
    "email": [
        "arjun.k@accenture.com",
        "meera.r@gmail.com",
        "riya.s@accenture.com"
    ]
})

df["email"] = df["email"].map(lambda x: "***@"+x.split("@")[1])

print(df)
```


Normalize a text column Easy
Use .apply() with a lambda to clean the status column: strip whitespace, lowercase, replace spaces with underscores.
df = pd.DataFrame({
    "pipeline": ["sales_etl","inventory","finance"],
    "status":   [" SUCCESS ", "Failed ", " IN PROGRESS"]
})

```python
df=pd.DataFrame({
    "pipeline": ["sales_etl","inventory","finance"], "status": [" SUCCESS ", "Failed ", " IN PROGRESS"]
})

df["status"] = df["status"].apply(lambda x: x.strip().lower().replace(" ","_"))

df
```

Apply a named function — revenue bucketing Medium
Write a named function revenue_tier(row) that uses both region and revenue to assign a tier: APAC + revenue > 100k → "Tier 1", any region + revenue > 50k → "Tier 2", else "Tier 3". Apply row-wise.
df = pd.DataFrame({
    "region":  ["APAC","EMEA","APAC","AMER","EMEA"],
    "revenue": [120000, 80000, 45000, 60000, 30000]
})

```python
df=pd.DataFrame({ "region": ["APAC","EMEA","APAC","AMER","EMEA"], 
"revenue": [120000, 80000, 45000, 60000, 30000] }
)

def revenue_tier(row):
    if row["region"] == "APAC" and row["revenue"] > 100000:
        return "Tier 1"
    
    elif row["region"] != "APAC" and row["revenue"]>50000:
        return "Tier 2"
    
    else:
        return "Tier 3"
    
df["tier"]=df.apply(revenue_tier, axis=1)

df
```

Row-wise apply — SLA breach flag Medium
Use .apply(axis=1) to add a sla_breached column. SLA is breached when status == "failed" OR duration_sec > 180.
df = pd.DataFrame({
    "pipeline":    ["sales_etl","inventory","finance","marketing"],
    "status":      ["success","failed","success","success"],
    "duration_sec":[120, 45, 210, 90]
})

```python
df=pd.DataFrame({
    "pipeline": ["sales_etl", "inventory", "finance", "marketing"],
    "status": ["success", "failed", "success", "success"],
    "duration_sec": [120, 45, 210, 90]
})

df["sla_breached"] = df.apply(
    lambda row: row["status"] == "failed" or row["duration_sec"] > 180,
    axis=1
)

df
```
