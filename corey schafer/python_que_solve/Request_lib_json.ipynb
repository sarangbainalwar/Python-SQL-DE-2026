```python
URL = "https://jsonplaceholder.typicode.com/posts/9999"

def check_url(url):
    sc=requests.get(url).status_code
    if sc == 200:
        print("URL is OK")
    elif sc == 404:
        print("URL not found")
    elif sc == 500:
        print("internal server error")
    else:
        print("Something else occured")
        
check_url(URL)
```

```python
import requests 
url = "https://jsonplaceholder.typicode.com/posts/1"
headers = {"User-Agent":"SarangBot/1.0"}
req=requests.Request("GET",url,headers=headers)
prepared=req.prepare()
print("Prepared URL: ")
print(prepared.url)

print("prepared headers: ")
for key,value in prepared.headers.items():
    print(f"{key}:{value}")

with requests.Session() as session:
    response=session.send(prepared)
    
print(response.status_code)
print(response.json())
```

```python
# # Intentional 404 Medium
# Fetch a URL that definitely returns 404 (/posts/99999). Use response.raise_for_status() and catch the exception. Print a clean error message without a traceback.
import requests
url = "https://jsonplaceholder.typicode.com/invaalid_end"

try:
    res=requests.get(url)
    print(res)
    res.raise_for_status()
    print(res.json())
except requests.exceptions.HTTPError as e:
    print(f"http error:{e}")
    
except requests.exceptions.RequestException as e:
    print(f"request error: {e}")
```
