```python
sentence = "Corey Schafer is awesome and has a great YouTube channel for learning Python!"

def func(s:str):
    maxi=0
    st={}
    for i in s:
        cnt = 0
        for j in i:
            cnt += 1
        st.update({i:cnt})
        maxi = max(maxi,cnt)
    return maxi,st
print(func(sentence.split()))
```

```python
n = 6

def fact(n:int):
    for i in range(1,n):
        n *= i
    return n
print(fact(n))
```

```python
s = "Corey Schafer is awesome and has a great YouTube channel for learning Python!"

def func(s:str):
    s=s.split()
    d={}
    for i in s:
        for j in i:
            if j in d:
                d[j]+=1
            else:
                d[j]=1
    return d
print(func(s))
```

```python
list1 = [48,587,6,5659,3569,959599]
def func(l:list):
    s = sum(l)
    a=s/len(l)
    m = max(l)
    n=min(l)
    return s,a,m,n
print(func(list1))
```

```python
# Write a function using **kwargs that prints employee details dynamically.
def emp_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
        
emp_details(name="John Doe", age=30, department="HR")
```
