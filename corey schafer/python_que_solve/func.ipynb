```python
def s(s:String):
    cnt = 0
    for i in s:
        if i in "aeiouAEIOU":
            cnt += 1
    return cnt
print(s("Corey Schafer"))
```

```python
sentence = "Corey Schafer is awesome and has a great YouTube channel for learning Python!"

def func(s:str):
    maxi=0
    st={}
    for i in s:
        cnt = 0
        for j in i:
            cnt += 1
        st.update({i:cnt})
        maxi = max(maxi,cnt)
    return maxi,st
print(func(sentence.split()))
```

```python
n = 6

def fact(n:int):
    for i in range(1,n):
        n *= i
    return n
print(fact(n))
```
