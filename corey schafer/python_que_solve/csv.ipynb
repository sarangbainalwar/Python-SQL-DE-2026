```python

# Write and read a pipeline log Easy
# Using csv.writer, write the data below to pipeline_log.csv. Then read it back using csv.reader and print each row. Confirm the header row is correct.

import csv

headers = ["pipeline_name", "status", "rows_processed", "run_date"]
rows = [
    ["sales_etl", "success", 95000, "2024-06-01"],
    ["inventory_sync", "failed", 0, "2024-06-01"],
    ["finance_load", "success", 5400, "2024-06-01"]
]

with open("pipeline_log.csv","w",newline="") as file:
    writer = csv.writer(file)
    writer.writerow(headers)
    writer.writerows(rows)
    
with open("pipeline_log.csv","r") as file:
    reader=csv.reader(file)
    data=list(reader)
    
    for row in data:
        print(row)
    
    if data[0] == headers:
        print("headers row is correct")
    else:
        print("headers row is incorrect")
```

```python
# DictReader — smarter row access Easy
# Read the same pipeline_log.csv you wrote in C1 using csv.DictReader. Print only the pipeline_name and status for rows where status is "failed".

import csv
with open("pipeline_log.csv","r") as file:
    reader = csv.DictReader(file)
    
    for row in reader:
        if row["status"] == "failed":
            print(row["pipeline_name"],row["status"])
```

```python
import csv 

transactions = [
    ["txn_id", "amount", "currency", "status"],
    ["T001", "4500.00", "INR", "settled"],
    ["T002", "120.50", "USD", "pending"],
    ["T003", "8900.00", "INR", "failed"]
]

with open("transactions.txt","w",newline="") as file:
    writer = csv.writer(file,delimiter="|") 
    writer.writerows(transactions)
    
with open("transactions.txt","r") as file:
    reader = csv.reader(file,delimiter="|")
    
    for row in reader:
        print(row)
```

```python
import csv

employees = [
    {"emp_id":"E01","name":"Arjun","department":"Engineering","salary":"85000"},
    {"emp_id":"E02","name":"Meera","department":"HR","salary":"60000"},
    {"emp_id":"E03","name":"Riya","department":"Engineering","salary":"65000"},
    {"emp_id":"E04","name":"Kiran","department":"Engineering","salary":"92000"},
    {"emp_id":"E05","name":"Sam","department":"Finance","salary":"71000"}
]

# Write employees to eng.csv
with open("eng.csv", "w", newline="") as file:
    fieldnames = ["emp_id", "name", "department", "salary"]
    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()
    writer.writerows(employees)

# Read, filter, and write to eng_filtered.csv
with open("eng.csv", "r") as infile, open("eng_filtered.csv", "w", newline="") as outfile:
    reader = csv.DictReader(infile)

    fieldnames = ["emp_id", "name", "department", "salary"]
    writer = csv.DictWriter(outfile, fieldnames=fieldnames)

    writer.writeheader()

    for row in reader:
        if row["department"] == "Engineering" and int(row["salary"]) > 70000:
            writer.writerow(row)
```

```python
with open("eng.csv", "r") as infile:
    reader = csv.DictReader(infile)
    
    dept_data={}
    
    total_salary_bill=0
    
    for row in reader:
        dept=row["department"]
        salary=int(row["salary"])
        
        total_salary_bill+=salary
        
        if dept not in dept_data:
            dept_data[dept]={
                "total_salary":0,
                "count":0
            }
            
        dept_data[dept]["total_salary"]+=salary
        dept_data[dept]["count"]+=1 
        
        avg_salary=total_salary_bill/sum(d["count"] for d in dept_data.values())
        
        with open("dept_summary.csv","w",newline="") as outfile:
            fieldnames=["department","total_salary","avg_salary","count"]
            
            writer=csv.DictWriter(outfile,fieldnames=fieldnames)
            writer.writeheader()
            
            for dept,data in dept_data.items():
                writer.writerow({
                    "department":dept,
                    "total_salary": data["total_salary"],
            "avg_salary": data["total_salary"] / data["count"],
            "count": data["count"]
                })
                
# Print overall statistics
print("Total Salary Bill:", total_salary_bill)
print("Average Salary:", avg_salary)
```
