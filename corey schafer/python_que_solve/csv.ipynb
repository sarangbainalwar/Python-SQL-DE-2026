```python

# Write and read a pipeline log Easy
# Using csv.writer, write the data below to pipeline_log.csv. Then read it back using csv.reader and print each row. Confirm the header row is correct.

import csv

headers = ["pipeline_name", "status", "rows_processed", "run_date"]
rows = [
    ["sales_etl", "success", 95000, "2024-06-01"],
    ["inventory_sync", "failed", 0, "2024-06-01"],
    ["finance_load", "success", 5400, "2024-06-01"]
]

with open("pipeline_log.csv","w",newline="") as file:
    writer = csv.writer(file)
    writer.writerow(headers)
    writer.writerow(rows)
```

```python
with open("pipeline_log.csv","r") as file:
    reader=csv.reader(file)
    data=list(reader)
    
    for row in data:
        print(row)
    
    if data[0] == headers:
        print("headers row is correct")
    else:
        print("headers row is incorrect")
```
