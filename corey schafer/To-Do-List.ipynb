```python
print("Welcome to your To-Do List!")

print("What would you like to do? (add: 1/delete: 2/view: 3/delete all: 4/exit: 0)")
action = int(input("Enter your choice: "))
while action != 0:
    if action == 1:
        with open("todo.txt","r") as file:
            max_index = 0
            for line in file:
                line = line.strip()
            
                num = line.split(".")[0]
                
                if num.isdigit():
                    max_index = max(max_index, int(num))
        index = max_index + 1
        task = input("Enter the task you want to add: ")
        with open("todo.txt","a") as file:
            file.write(f"{index}. {task}\n")
        print("Task added successfully!")
        action = int(input("What would you like to do next? (add: 1/delete: 2/view: 3/delete all: 4/exit: 0)"))
        
    elif action == 2:
        print("would you like to delete a specific task ? (yes: 1/no: 0)")
        delete_choice = int(input("Enter your choice: "))   
        if delete_choice == 1:
            with open("todo.txt","r") as file:
                lines = file.readlines()
                if not lines:
                    print("Your To-Do List is empty.")
                else:
                    print("Your To-Do List:")
                    for line in lines:
                        print(line.strip())
            task_number = int(input("Enter the index of the task you want to delete: "))
            with open("todo.txt","w") as file:
                for line in lines:
                    if not line.startswith(f"{task_number}."):
                        file.write(line)
        elif delete_choice == 0:
            continue
         
        action = int(input("What would you like to do next? (add: 1/delete: 2/view: 3/delete all: 4/exit: 0)"))
        
        
    elif action == 3:
        with open("todo.txt","r") as file:
            lines = file.readlines()
            if not lines:
                print("Your To-Do List is empty.")
            else:
                print("Your To-Do List:")
                for line in lines:
                    print(line.strip())
        action = int(input("What would you like to do next? (add: 1/delete: 2/view: 3/delete all: 4/exit: 0)"))
        
    elif action == 4:
        print("would you like to delete all tasks ? (yes: 1/no: 0)")
        delete_choice = int(input("Enter your choice: "))
        if delete_choice == 1:
            with open("todo.txt","w") as file:
                file.write("")
            print("All tasks have been deleted.")
        elif delete_choice == 0:
            continue
        action = int(input("What would you like to do next? (add: 1/delete: 2/view: 3/delete all: 4/exit: 0)"))
        
    else:
        print("Invalid choice. Please try again.")
        action = int(input("What would you like to do next? (add: 1/delete: 2/view: 3/delete all: 4/exit: 0)"))
        
print("Goodbye!, you have exited the To-Do List.")
```
