```python
class DataPipeline:
    company = "Accenture"
    pcnt=0
    
    def __init__(self,name,source,target):
        self.name = name
        self.source=source
        self.target=target
        self.status="idle"
        self.rows_processed=0
        DataPipeline.pcnt+=1
    
    def run(self,rows):
        self.rows_processed=rows
        self.status="success"
        print(f"pipeline '{self.name}' ran: {rows} rows loaded")
        
    def reset(self):
        self.rows_processed=0
        self.status="idle"
                
    def info(self):
        print(f"{self.name} | {self.source} | {self.status}")
```

```python
p1=DataPipeline("sales_etl","BQ","MySQL")
p2 = DataPipeline("inventory_sync","S3","Redshift")
```

```python
p1.run(95000)
p2.run(12000)
```

```python
p1.info()
p2.info()
```
