```python
class databaseconnection:
    def __init__ (self,host,user,password):
        self.host=host # public
        self._user=user # protected
        self.__password = password # private-name mangled 
        self.__connected = False
    
    def connect(self):
        print(f"Connecting to {self.host} as {self._user}...")
        self.__connected=True
        print("connected")
        
    def disconnected(self):
        self.__connected=False
        print("disconnected")
        
    def is_connected(self):
        return self.__connected
    
    def get_connection_info(self):
        return {"host": self.host,"user":self._user,"connected":self.__connected}
    
    def change_password(self,old,new):
        if old == self.__password:
            self.__password = new
            print("password changed")
        else:
            print("wrong password")
```

```python
df=databaseconnection("prod.db.internal","etl_user","sarang")
```

```python
df.connect()
```

```python
print(df.is_connected())
print(df.get_connection_info())
```
