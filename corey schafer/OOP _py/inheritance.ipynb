```python
class datapipeline:
    def __init__(self,name,source):
        self.name=name
        self.source=source
        self.status="idle"
    
    def run(self):
        print(f"running {self.name} from {self.source}")
        
    def info(self):
        print(f"[{self.status}] {self.name} <- {self.source}")
    
class bigquerypipeline(datapipeline):
    def __init__(self,name,dataset,table):
        super().__init__(name,source="bigquery")
        self.dataset=dataset
        self.table=table
    
    def run(self):
        query = f"select * from {self.dataset}.{self.table}"
        print(f"BQ query:{query}")
        self.status="success"
        
    def export_schema(self):
        print(f"exporting schema for {self.dataset}.{self.table}")
        
class s3pipeline(datapipeline):
    def __init__(self,name,bucket,file_format):
        super().__init__(name,source="s3")
        self.bucket = bucket
        self.file_format=file_format
        
    def run(self):
        print(f"reading {self.file_format} files from s3://{self.bucket}")
        self.status="success"

bq=bigquerypipeline("sales","raw_data","orders")
s3=s3pipeline("inventory","acc_data","parquet")
```

```python
bq.run()
```

```python
bq.info()
```

```python
bq.export_schema()
```

```python
s3.run()
```

```python
print(isinstance(bq,datapipeline))
print(isinstance(bq,s3pipeline))
```
