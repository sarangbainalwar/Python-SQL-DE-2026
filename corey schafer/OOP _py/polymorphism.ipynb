```python
class DataPipeline:
    def __init__(self,name):self.name=name
    def extract(self): raise NotImplementedError
    def load(self): raise NotImplementedError
    
class bqpipeline(DataPipeline):
    def extract(self):
        print(f"[{self.name}] extractng via bq sql...")
    def load(self):
        print(f"[{self.name}] loading to bigquery table...")

class s3pipeline(DataPipeline):
    def extract(self):
        print(f"[{self.name}] scanning s3 bucket for new files")
    def load(self):
        print(f"[{self.name}] weiting parquet to s3...")
        
class mysqlpipeline(DataPipeline):
    def extract(self):
        print(f"[{self.name}] querying mysql with JDBC...")
    def load(self):
        print(f"[{self.name}] bulk inserting into mysql...")
        
class kafkapipeline(DataPipeline):
    def extract(self):
        print(f"[{self.name}] kafka")
    def load(self):
        print(f"[{self.name}] kafka")
```

```python
pipelines=[
    bqpipeline("sales"),
    s3pipeline("inventory"),
    mysqlpipeline("finance"),
    kafkapipeline("Data Engineer")
]

for p in pipelines:
    p.extract()
    p.load()
    print()
```
