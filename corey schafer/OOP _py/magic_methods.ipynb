```python
class Dataset:
    def __init__(self, name, records):
        self.name = name
        self.records = records       # list of dicts

    def __str__(self):               # print(dataset) uses this
        return f"Dataset('{self.name}', {len(self.records)} rows)"

    def __repr__(self):              # REPL / debugging uses this
        return f"Dataset(name='{self.name}', rows={len(self.records)})"

    def __len__(self):               # len(dataset)
        return len(self.records)

    def __eq__(self, other):         # dataset1 == dataset2
        return self.name == other.name and len(self) == len(other)

    def __lt__(self, other):         # dataset1 < dataset2 (by row count)
        return len(self) < len(other)

    def __add__(self, other):        # dataset1 + dataset2 = merged
        merged = self.records + other.records
        return Dataset(f"{self.name}+{other.name}", merged)

    def __contains__(self, key):    # "field" in dataset
        return any(key in row for row in self.records)


sales = Dataset("sales", [
    {"id":1,"amount":4500}, {"id":2,"amount":3200}
])
returns = Dataset("returns", [
    {"id":3,"amount":-800}
])

print(sales)                    # Dataset('sales', 2 rows)
print(len(sales))               # 2
print(sales < returns)          # False — sales has more rows
combined = sales + returns
print(combined)                 # Dataset('sales+returns', 3 rows)
print("amount" in sales)        # True
```
