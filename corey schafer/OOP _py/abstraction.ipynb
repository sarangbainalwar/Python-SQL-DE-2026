```python
from abc import ABC,abstractmethod
```

```python
class DataConnector(ABC): #abstract base class
    def __init__(self,name):
        self.name=name
        
    @abstractmethod
    def connect(self): pass
    
    @abstractmethod
    def read(self,query): pass
    
    @abstractmethod
    def write(self,data): pass
    
    def disconnect(self): print(f"[{self.name}] disconnected") #concrete method shared by all
```

```python
class bqconnector(DataConnector):
    def connect(self):
        print(f"[{self.name}] authenticating the GCP service account")
    
    def read(self,query):
        print(f"[{self.name}] executing: {query}")
        return [{"id":1,"amount":4500}]
    
    def write(self, data):
        print(f"[{self.name}] Writing {len(data)} rows to BQ table...")
```

```python
class MySQLConnector(DataConnector):
    def connect(self):
        print(f"[{self.name}] Opening MySQL connection pool...")
    def read(self, query):
        print(f"[{self.name}] MySQL query: {query}")
        return [{"id": 2, "amount": 3200}]
    def write(self, data):
        print(f"[{self.name}] Bulk INSERT {len(data)} rows to MySQL...")
```

```python
bq = bqconnector("BQ-Prod")
bq.connect()
rows = bq.read("SELECT * FROM sales.orders LIMIT 100")
bq.write(rows)
bq.disconnect()    # inherited concrete method
```
