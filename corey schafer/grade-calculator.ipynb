```python
print("Welcome to grade calculator")
total_subs = int(input("How many subjects do you have? "))
total_grade = 0
for i in range(total_subs):
    marks = float(input(f"Enter marks for subject {i+1}: "))
    total_grade = total_grade + marks
total_grade = total_grade / total_subs

print(f"Your average grade is: {total_grade}")
```
