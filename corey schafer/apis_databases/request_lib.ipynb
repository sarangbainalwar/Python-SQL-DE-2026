```python
# ============================================================
# WEEK 7 - DAY 2 (continued): Nested JSON & Field Extraction
# ============================================================

import requests
import pandas as pd

users = requests.get("https://jsonplaceholder.typicode.com/users").json()


# ─────────────────────────────────────────────────────────────
# SECTION 3: Flattening Nested JSON — 3 Approaches
# ─────────────────────────────────────────────────────────────

# --- Approach A: Manual extraction (most control) ---
print("=== Approach A: Manual extraction ===")
flat_records = []
for user in users:
    flat_records.append({
        "id"          : user["id"],
        "name"        : user["name"],
        "email"       : user["email"],
        "city"        : user["address"]["city"],       # nested dict
        "zipcode"     : user["address"]["zipcode"],    # nested dict
        "company_name": user["company"]["name"],       # nested dict
        "lat"         : user["address"]["geo"]["lat"], # 2 levels deep
        "lng"         : user["address"]["geo"]["lng"],
    })

df_flat = pd.DataFrame(flat_records)
print(df_flat[["id", "name", "city", "company_name"]].head())


# --- Approach B: pd.json_normalize (built-in pandas tool) ---
print("\n=== Approach B: pd.json_normalize ===")
df_normalized = pd.json_normalize(
    users,
    sep="_"   # nested keys become: address_city, address_geo_lat etc.
)
print(f"Columns after normalize: {list(df_normalized.columns)}")
print(df_normalized[["id", "name", "address_city", "company_name"]].head())


# --- Approach C: Nested list inside JSON (e.g. tags, items arrays) ---
print("\n=== Approach C: Exploding nested lists ===")

# Simulate API response with nested list
sample = [
    {"id": 1, "name": "Alice", "skills": ["python", "sql", "spark"]},
    {"id": 2, "name": "Bob",   "skills": ["java", "kafka"]},
    {"id": 3, "name": "Carol", "skills": ["python", "airflow", "dbt"]},
]

df_nested = pd.DataFrame(sample)
print("Before explode:")
print(df_nested)

df_exploded = df_nested.explode("skills").reset_index(drop=True)
print("\nAfter explode (one row per skill):")
print(df_exploded)


# ─────────────────────────────────────────────────────────────
# SECTION 4: Extracting Specific Fields Safely
# ─────────────────────────────────────────────────────────────
print("\n=== Safe Field Extraction ===")

# UNSAFE — will crash if key is missing
# name = user["missing_key"]  # KeyError!

# SAFE — .get() returns None instead of crashing
for user in users[:3]:
    name    = user.get("name", "UNKNOWN")
    phone   = user.get("phone", "NO PHONE")      # may be missing in real APIs
    website = user.get("website", "NO WEBSITE")
    city    = user.get("address", {}).get("city", "NO CITY")  # nested safe get

    print(f"  {name} | {phone} | {city}")


# ─────────────────────────────────────────────────────────────
# SECTION 5: Real-world Pattern — API → Clean DataFrame → Export
# ─────────────────────────────────────────────────────────────
print("\n=== Full Pattern: API → DataFrame → CSV ===")

# 1. Fetch
posts  = requests.get("https://jsonplaceholder.typicode.com/posts").json()
users  = requests.get("https://jsonplaceholder.typicode.com/users").json()

# 2. Flatten to DataFrames
df_posts = pd.DataFrame(posts)
df_users = pd.DataFrame(users)[["id", "name", "email"]]

# 3. Join (like SQL JOIN)
df_merged = df_posts.merge(
    df_users,
    left_on="userId",
    right_on="id",
    suffixes=("_post", "_user")
)

# 4. Select and rename
df_final = df_merged[["id_post", "title", "name", "email"]].rename(columns={
    "id_post": "post_id",
    "name"   : "author_name",
    "email"  : "author_email",
})

print(f"Final shape: {df_final.shape}")
print(df_final.head())

# 5. Export
# df_final.to_csv("/home/claude/posts_with_authors.csv", index=False)
print("\nSaved to posts_with_authors.csv ✓")

print("\n✓ Day 2 complete — JSON APIs to DataFrames covered")
```
