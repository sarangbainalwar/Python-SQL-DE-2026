```python
import json

people_string = '''
{
  "people": [
    {
      "name": "John Smith",
      "phone": "615-555-7164",
      "emails": [
        "johnsmith@bogusemail.com",
        "john.smith@place.com"
      ],
      "has_license": false
    },
    {
      "name": "Jane Doe",
      "phone": "560-555-5153",
      "emails": null,
      "has_license": true
    }
  ]
}
'''

data = json.loads(people_string)
print(data)
```

```python
for person in data['people']:
    print(person['name'])
```

```python
for person in data['people']:
    del person['phone']
```

```python
new_string = json.dumps(data, indent = 2,sort_keys=True)
print(new_string)
```

```python
with open('states.json') as f:
    data = json.load(f)
```

```python
for state in data['states']:
    print(state['name'],state['abbreviation'])
```

```python
for state in data['states']:
    del state['capital']
```

```python
with open('new_states.json','w') as f:
    json.dump(data,f,indent=2)
```
