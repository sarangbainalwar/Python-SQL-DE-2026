```python
import pandas as pd

query="""
select * from employee_new;
"""

df=pd.read_sql(query,engine)
print(df)
```

```python
query="""select * from employee_new where salary > 60000"""

df=pd.read_sql(query,engine)
df
```

```python
# NOTE Write DataFrame to Database

df=pd.DataFrame({
    "employee_id":[6,7],
    "name":["Rohit","Kiran"],
    "department":["IT","HR"],
    "salary":[75000,80000]
})

df.to_sql(
    "employee_new",
    engine,
    index=False
)
```

```python
df.to_sql(
    "employee_new",
    engine,
    if_exists="replace"
)
```

```python
df.to_sql(
    "employee_new",
    engine,
    if_exists="append"
)
```
