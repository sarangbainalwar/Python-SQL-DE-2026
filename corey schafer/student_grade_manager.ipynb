```python
master_li = {}
```

```python
print("Enter name of student to enter marks: ")
name=str(input())
total_subs = int(input(f"input total subs for {name}"))
sub_marks = {}
for i in range(total_subs):
    sub = str(input("input name of subject: "))
    marks=int(input(f"enter marks for subject {sub}"))
    sub_marks[sub] = marks

master_li[name] = sub_marks

print(master_li)
```
